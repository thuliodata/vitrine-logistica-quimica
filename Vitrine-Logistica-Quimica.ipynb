```python
# 1. Execute esta linha no Colab/Jupyter para instalar o Gradio:
!pip install gradio pandas

# 2. Em seguida, execute este código:
import gradio as gr
import pandas as pd

# Banco de dados simulado
pedidos = [
    {
        "ID": 1,
        "Coleta": "Guarulhos/SP",
        "Entrega": "Ribeirão Preto/SP",
        "Volume": 2,
        "Embalagem": "IBC",
        "Cubagem": "2,4 m³",
        "Status": "Disponível para coleta",
        "Transportadora Sugerida": "TransQuímica Express",
        "E-mail": "contato@transquimica.com",
        "Melhor Frete": "R$ 1.850,00"
    }
]

def cadastrar_pedido(coleta, entrega, volume, embalagem, cubagem, transportadora, email, frete):
    novo = {
        "ID": len(pedidos) + 1,
        "Coleta": coleta,
        "Entrega": entrega,
        "Volume": volume,
        "Embalagem": embalagem,
        "Cubagem": cubagem,
        "Status": "Disponível para coleta",
        "Transportadora Sugerida": transportadora,
        "E-mail": email,
        "Melhor Frete": frete
    }
    pedidos.append(novo)
    return pd.DataFrame(pedidos), "✅ Pedido cadastrado e publicado na Vitrine com sucesso!"

def listar_pedidos():
    return pd.DataFrame(pedidos)

def solicitar_cotacao(id_pedido):
    for p in pedidos:
        if str(p["ID"]) == str(id_pedido):
            return f"📧 E-mail de solicitação de cotação enviado para {p['E-mail']} (Transportadora: {p['Transportadora Sugerida']})!"
    return "Pedido não encontrado."

# Interface do Gradio
with gr.Blocks(title="Vitrine Logística - Setor Químico") as app:
    gr.Markdown("# 🧪 Cargas Disponíveis - Setor Químico")
    gr.Markdown("Sistema de Alocação de Fretes")

    with gr.Tab("📦 Coletas/Entregas Disponíveis"):
        tabela = gr.Dataframe(value=pd.DataFrame(pedidos), interactive=False)
        btn_atualizar = gr.Button("🔄 Atualizar")
        btn_atualizar.click(fn=listar_pedidos, outputs=tabela)

        gr.Markdown("### 📤 Solicitar Cotação / Coleta")
        input_id = gr.Textbox(label="ID do Pedido", value="1")
        btn_cotar = gr.Button("Enviar Solicitação para a Transportadora")
        output_cotacao = gr.Textbox(label="Status do Envio")
        btn_cotar.click(fn=solicitar_cotacao, inputs=input_id, outputs=output_cotacao)

    with gr.Tab("➕ Cadastrar Novo Pedido"):
        with gr.Row():
            coleta = gr.Textbox(label="Local de Coleta", value="Guarulhos/SP")
            entrega = gr.Textbox(label="Local de Entrega", value="Ribeirão Preto/SP")
        with gr.Row():
            volume = gr.Number(label="Volume", value=2)
            embalagem = gr.Dropdown(choices=["IBC", "Tambor", "Granel", "Bombonas"], label="Embalagem", value="IBC")
            cubagem = gr.Textbox(label="Cubagem", value="2,4 m³")
        with gr.Row():
            transportadora = gr.Textbox(label="Transportadora Sugerida", value="TransQuímica Express")
            email = gr.Textbox(label="E-mail da Transportadora", value="contato@transquimica.com")
            frete = gr.Textbox(label="Melhor Frete Estimado", value="R$ 1.850,00")

        btn_salvar = gr.Button("Salvar Pedido")
        status_cadastro = gr.Textbox(label="Resultado")
        btn_salvar.click(
            fn=cadastrar_pedido,
            inputs=[coleta, entrega, volume, embalagem, cubagem, transportadora, email, frete],
            outputs=[tabela, status_cadastro]
        )

app.launch(share=True)
```
